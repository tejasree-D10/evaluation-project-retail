```sql
%sql

-- =====================================
-- USE CATALOG & SCHEMA
-- =====================================

USE CATALOG `retail-dwh-project`;
USE SCHEMA bronze;


-- =====================================
-- DROP OLD CONFLICTING TABLES
-- =====================================

```

```sql
%sql

CREATE OR REPLACE TABLE bronze.customers_raw AS
SELECT *
FROM read_files(
  "s3://retail-etl-dwh-lakehouse/raw/customers_src_*.csv",
  format => "csv",
  header => true,
  inferSchema => true
);
```

```sql
%sql

-- =====================================
-- CREATE PRODUCTS RAW TABLE
-- =====================================

CREATE OR REPLACE TABLE bronze.products_raw AS
SELECT *
FROM read_files(
  "s3://retail-etl-dwh-lakehouse/raw/products_src_*.csv",
  format => "csv",
  header => true,
  inferSchema => true
);


-- =====================================
-- CREATE SALES RAW TABLE
-- =====================================
```

```sql

CREATE OR REPLACE TABLE bronze.sales_raw AS
SELECT *
FROM read_files(
  "s3://retail-etl-dwh-lakehouse/raw/sales_transactions_src_*.csv",
  format => "csv",
  header => true,
  inferSchema => true
);
```

```sql

CREATE OR REPLACE TABLE bronze.stores_raw AS
SELECT *
FROM read_files(
  "s3://retail-etl-dwh-lakehouse/raw/stores_src_*.csv",
  format => "csv",
  header => true,
  inferSchema => true
);
```

```sql
%sql

SELECT 'customers_raw' AS table_name, COUNT(*) AS total_rows
FROM bronze.customers_raw

UNION ALL

SELECT 'products_raw' AS table_name, COUNT(*) AS total_rows
FROM bronze.products_raw

UNION ALL

SELECT 'sales_raw' AS table_name, COUNT(*) AS total_rows
FROM bronze.sales_raw

UNION ALL

SELECT 'stores_raw' AS table_name, COUNT(*) AS total_rows
FROM bronze.stores_raw;
```
